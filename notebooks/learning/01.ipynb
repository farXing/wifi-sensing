```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(13)

x = np.random.uniform(1,10,100)
y = 3*x + 2 + np.random.normal(0,1,100)
```

```python
def predict(x,w,b):
  return x*w + b
```

```python
x3 = np.array([1,2,3])
y3 = np.array([5,8,11])

print(predict(x3,3,2))
```

```python
def mse(y,y_true):
  return np.mean((y - y_true)**2)
```

```python
def gradient(x,y,w,b):
  errors = predict(x,w,b) - y
  dw = 2 * np.mean(errors*x)
  db = 2 * np.mean(errors)
  return dw, db
```

```python
def train(x,y,lr,steps):
  w, b = 0.0,0.0
  history = []
  for i in range(steps):
    dw, db = gradient(x,y,w,b)
    w = w - lr*dw
    b = b - lr*db
    history.append(mse(predict(x,w,b), y))
  return w,b,history
```

```python
w,b,fl = train(x,y,0.001,10000)
print(w,b,fl[-1])
```

```python
plt.plot(fl[50:])
plt.title('Ошибка по шагам')
plt.show()

xs = np.linspace(1,10,100)
plt.scatter(x, y, s=10)
plt.plot(xs, predict(xs, w, b), color='red')
plt.title('Что нашла модель')
plt.show()
```
